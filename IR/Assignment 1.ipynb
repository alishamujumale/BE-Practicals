```python
# 1. Write a program for pre-processing of a text document such as stop word removal, stemming.

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

#install only once when running for the first time
nltk.download('punkt')
nltk.download('stopwords')

# document = input("Enter your document: ")
document = "Machine Learning cannot be mastered in a two months it needs years of practicing and studying"

#Tokenization
tokenizedDocument = word_tokenize(document.lower())
print("------After Tokenization-------")
print(tokenizedDocument,"\n")

#Stop-word Removal
stopWords = set(stopwords.words('english'))
filteredDocument = []
for word in tokenizedDocument:
    if word in stopWords:
        continue
    filteredDocument.append(word)
print("------After Stopword Removal------")
print(filteredDocument,"\n")

#Stemming
stemmer = PorterStemmer()
stemmedWords = []
for word in filteredDocument:
    stemmedWord = stemmer.stem(word)
    stemmedWords.append(stemmedWord)
print("------After Stemming-------")
print(stemmedWords,"\n")

```
