```python
# Molecular Docking and Virtual Screening
# Sample Data

import pandas as pd

# Create sample docking data
data = {
    "Ligand": [
        "Ligand_A",
        "Ligand_B",
        "Ligand_C",
        "Ligand_D",
        "Ligand_E",
        "Ligand_F"
    ],
    
    "H_Bonds": [3, 2, 4, 1, 3, 2],
    
    "Hydrophobic_Interactions": [5, 4, 6, 3, 7, 5],
    
    "Molecular_Weight": [310, 285, 340, 250, 360, 300],
    
    "Binding_Score": [-7.8, -6.4, -8.5, -5.2, -9.1, -7.0]
}

# Create DataFrame
df = pd.DataFrame(data)

print("Virtual Screening Dataset:")
print(df)

# Rank ligands based on binding score
result = df.sort_values(
    by="Binding_Score",
    ascending=True
)

print("\nRanked Ligands:")
print(result)

# Best ligand
best = result.iloc[0]

print("\nBest Potential Lead Compound:")
print("Ligand:", best["Ligand"])
print("Binding Score:", best["Binding_Score"], "kcal/mol")
```

```python
import matplotlib.pyplot as plt

# Convert negative binding scores to positive values for visualization
df["Binding_Affinity"] = abs(df["Binding_Score"])

plt.bar(df["Ligand"], df["Binding_Affinity"])

plt.xlabel("Ligands")
plt.ylabel("Binding Affinity (absolute kcal/mol)")
plt.title("Virtual Screening Results")

plt.show()
```
